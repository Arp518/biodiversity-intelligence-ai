# Scientific Document Text Extraction

The scientific PDF documents are extracted page-by-page while preserving
source and page metadata.

This prepares the scientific corpus for later chunking, embedding, retrieval,
and citation verification.

Each extracted page retains:
- source ID
- original filename
- page number
- extracted text

No chunking or embeddings are created during this step.

```python
import fitz

print("PyMuPDF loaded successfully")
```

```python
from pathlib import Path

knowledge_dir = Path("../data/knowledge")

pdf_path = knowledge_dir / "SRC04_Cover crop species, with a special focus on legumes.pdf"

print("File exists:", pdf_path.exists())
print("Path:", pdf_path)
```

```python
doc = fitz.open(pdf_path)

print("Number of pages:", len(doc))

metadata = doc.metadata

print("\nPDF metadata:")
for key, value in metadata.items():
    print(f"{key}: {value}")
```

```python
pages = []

for page_index in range(len(doc)):
    page = doc[page_index]
    text = page.get_text("text").strip()

    pages.append({
        "source_id": "SRC04",
        "filename": pdf_path.name,
        "page_number": page_index + 1,
        "text": text
    })

doc.close()

print("Pages extracted:", len(pages))
```

```python
for page in pages:
    print(
        f'{page["source_id"]} | '
        f'Page {page["page_number"]} | '
        f'{len(page["text"])} characters'
    )
```

```python
empty_pages = [
    page["page_number"]
    for page in pages
    if not page["text"].strip()
]

print("Empty pages:", empty_pages)
```

## Automated Corpus Extraction

After validating text extraction on SRC04, the same extraction process is
applied to all scientific documents registered in `sources.csv`.

Each extracted page retains its source ID, filename, page number, and text so
that provenance can be preserved during later chunking and retrieval.

```python
import pandas as pd
import fitz
from pathlib import Path

knowledge_dir = Path("../data/knowledge")

sources_df = pd.read_csv(knowledge_dir / "sources.csv")

def extract_pdf_pages(pdf_path, source_id):
    pages = []

    with fitz.open(pdf_path) as doc:
        for page_index, page in enumerate(doc):
            text = page.get_text("text").strip()

            pages.append({
                "source_id": source_id,
                "filename": pdf_path.name,
                "page_number": page_index + 1,
                "text": text
            })

    return pages
```

```python
all_pages = []

for _, source in sources_df.iterrows():

    source_id = source["source_id"]
    filename = source["local_filename"]

    pdf_path = knowledge_dir / filename

    if not pdf_path.exists():
        print(f"MISSING: {source_id} | {filename}")
        continue

    extracted_pages = extract_pdf_pages(pdf_path, source_id)

    all_pages.extend(extracted_pages)

    print(
        f"{source_id} | "
        f"{len(extracted_pages)} pages extracted | "
        f"{filename}"
    )

print("\nTotal pages extracted:", len(all_pages))
```
